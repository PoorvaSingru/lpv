```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Load data
df = pd.read_csv(r"C:\Users\LENOVO\OneDrive\Desktop\downloads\letter-recognition.csv")

# Split features & target
X = df.iloc[:, 1:].values
y = df.iloc[:, 0].values   # letters A-Z

# Encode labels (A-Z → 0-25)
y = LabelEncoder().fit_transform(y)

# Split
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Scale
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

# Model
model = Sequential([
    Dense(64, activation='relu', input_shape=(16,)),
    Dense(32, activation='relu'),
    Dense(26, activation='softmax')   # 26 classes
])

model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Train
model.fit(x_train, y_train, epochs=10, verbose=0)

# Evaluate
print("Accuracy:", model.evaluate(x_test, y_test)[1])
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, model.predict(x_test).argmax(axis=1)))
```
